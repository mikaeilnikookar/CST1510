## 1. Creating a list, and indexing

Indexing counts from 0 — the same rule as `range()`.

````python
labels = ["R-001", "R-002", "R-003"]

print(labels[0])
print(labels[1])
print(labels[2])
````

```
index:     0        1        2
labels:  ["R-001", "R-002", "R-003"]
            ▲                  ▲
        labels[0]          labels[-1]
```

## 2. Negative indexing

Counts from the end. `labels[-1]` is the last item, whatever the length turns out
to be — no need to know how many items there are.

````python
print(labels[-1])   # R-003
print(labels[-2])   # R-002
print(len(labels))   # 3
````

## 3. Slicing

Same rule as `range(start, stop)` — `stop` is never included.

````python
values = [10, 20, 30, 40, 50]

print(values[1:3])   # [20, 30]
print(values[:2])     # [10, 20]
print(values[2:])     # [30, 40, 50]
````

## 4. Mutating and looping

Lists can be changed in place — strings cannot. `for` loops directly over the
values, no counting needed.

````python
values[0] = 99
values.append(60)
values.remove(20)
print(values)

for v in values:
    print(v)
````

unpacked it into three names.

````python
point = (3, 7)
print(point[0])   # 3 - indexing works the same as a list

# point[0] = 5     # TypeError: 'tuple' object does not support item assignment
````

## 6. Strings slice the same way too

You have been indexing and slicing without knowing it since Week 1. A string
follows the exact same rules as a list — and like a tuple, it is locked.

````python
record_id = "R-001"

print(record_id[0])      # 'R' - same indexing rule
print(record_id[-1])      # '1' - same negative indexing
print(record_id[0:2])      # 'R-' - same slicing rule, stop excluded
print(record_id[-3:])       # '001'
````

```
# record_id[0] = "X"
# TypeError: 'str' object does not support item assignment
```

Same lock as a tuple — you can read a character, not change one in place. The
full toolbox of string methods (`.upper()`, `.strip()`, `.replace()` and more) is
coming properly in Week 7. For now: the shape you already know for lists and
tuples works on text too.

## 7. Building a list in one line — list comprehensions

A `for` loop that builds a new list by appending is common enough that Python
has a shorter way to write exactly that.

````python
doubled = []
for v in values:
    doubled.append(v * 2)
print(doubled)

# the same thing, written as a list comprehension
doubled = [v * 2 for v in values]
print(doubled)
````

```
[  v * 2   for v in values  ]
     │            │
what to keep   where each v comes from
```

Add `if condition` at the end to filter as you build.

````python
big_values = [v for v in values if v >= 50]
print(big_values)
````

## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| `labels[3]` when there are only 3 items | `IndexError: list index out of range` | the last valid index is `len(labels) - 1` |
| `values[1:3]` expecting 3 items | gives 2 | `stop` is never included, same as `range()` |
| trying `name[0] = "z"` on a string | `TypeError` | strings cannot be changed this way — lists can |
| `.remove(x)` when `x` is not in the list | `ValueError` | check with `in` first, or catch the error |
| `point[0] = 5` on a tuple | `TypeError: 'tuple' object does not support item assignment` | tuples cannot be changed — use a list if you need to |
| a comprehension with `if`/`else` logic *and* a filter, tangled together | hard to read, easy to get wrong | if it needs more than one clear idea, write the `for` loop instead |
